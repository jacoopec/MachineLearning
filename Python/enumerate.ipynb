An iterable in Python is any object you can go through one element at a time, typically with a for loop.

The enumerate() function in Python is used to iterate over an iterable while keeping track of both the index and the value.
It returns pairs in the form (index, element). 
This removes the need to manually maintain a counter variable during iteration.

```
# list
[1, 2, 3]

# tuple
(1, 2, 3)

# string
"hello"

# set
{1, 2, 3}

# dictionary
{"a": 1, "b": 2}

# range
range(5)
```

````python
numbers = [10, 20, 30]

for x in numbers:
    print(x)
````

````python
# An iterable is something you can ask to produce an iterator:
# The iterator then produces values one at a time using next():

it = iter(numbers)

print(next(it))  # 10
print(next(it))  # 20
print(next(it)) 
````

````python
# Conceptually, a for loop does something similar internally:
while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
````

````python
a = ["thing1","thing2","thing3"]

r = list(enumerate(a))
````

````python
for i,v in enumerate(a):
    print(i,v)
````

````python
dictionary = {"a": 41, "b": 42, "c": 85}

for i, (k,v) in enumerate(dictionary):
    print(i,k,v)
    
````

A **generator** in Python is a special kind of iterator that produces values one at a time, only when they are needed.

The key idea is that generators are lazy: they do not usually create and store all their values in memory at once.
A generator function looks like a normal function, but uses yield instead of return:

````python
def count_up_to(n):
    i = 1

    while i <= n:
        yield i
        i += 1
        
        
# The important difference between yield and return is that yield pauses the function instead of terminating it.
````

````python
# Calling it does not immediately execute the whole function:
# g is a generator object.
g = count_up_to(3)
````

````python
# You can retrieve values using next():

print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3

````

Suppose you want one million numbers.
A list creates all of them immediately:
```numbers = [x for x in range(1_000_000)]```
A generator produces them only when requested:
```numbers = (x for x in range(1_000_000))```

````python
# List comprehension
[x * 2 for x in range(10)]

# Generator expression
(x * 2 for x in range(10))
````

Generators are iterators
Every generator is an iterator, and therefore also an iterable.
