```python
# print(df["country"].value_counts())

import matplotlib.pyplot as plt
import pandas as pd

# Select the 2021 unemployment column
data = df["laps"].dropna()

# Create the histogram
counts, bins, patches = plt.hist(data, bins=20, edgecolor='black')
print(patches[0].get_x())
print(patches[0].get_width())
print(patches)

# Convert frequencies to percentages
total = counts.sum()
percentages = (counts / total) * 100

# Add percentage labels on each bar
for count, percentage, patch in zip(counts, percentages, patches):
    x = patch.get_x() + patch.get_width() / 2
    y = patch.get_height()
    plt.text(x, y, f"{percentage:.1f}%", ha="center", va="bottom")

plt.title("2021 Unemployment – Histogram with Percentages")
plt.xlabel("Unemployment (%)")
plt.ylabel("Frequency")

plt.show()
```
