# 🧠 Adult Income Dataset – Feature Engineering

```
import pandas as pd

# Load dataset
columns = [
    "age", "workclass", "fnlwgt", "education", "education-num",
    "marital-status", "occupation", "relationship", "race", "sex",
    "capital-gain", "capital-loss", "hours-per-week", "native-country", "income"
]

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
df = pd.read_csv(url, names=columns, header=None, sep=",\s*", engine='python')

# Clean strings and handle missing values
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
df.replace('?', pd.NA, inplace=True)
df.dropna(inplace=True)
df.head()
```

## 🎯 Step 1: Drop Unnecessary Columns

```
# Drop 'education' (use 'education-num') and 'fnlwgt' (not informative)
df = df.drop(columns=['education', 'fnlwgt'])
df.head()
```

## 🔄 Step 2: Encode Categorical Variables

```
# One-hot encode categorical features
df_encoded = pd.get_dummies(df, drop_first=True)
df_encoded.head()
```

## 🎯 Step 3: Encode Target Variable

```
# Encode 'income' to binary
df_encoded['income_>50K'] = df_encoded['income_>50K'].astype(int)
y = df_encoded['income_>50K']
X = df_encoded.drop(columns=['income_>50K'])

print("X shape:", X.shape)
print("y value counts:")
print(y.value_counts())
```
