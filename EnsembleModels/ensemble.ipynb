# What is the probability that the ensemble prediction is correct, given the accuracies of its individual models?

For a simple ensemble with majority voting, you can compute it using binomial probability, assuming the models are independent and have the same accuracy.

P = Σ C(n, k) p^k (1 - p)^(n-k)





```python
number_of_models = 5
accuracy_of_each_model = 0.75

# For majority voting, the ensemble is correct if more than half of the models are correct.

# Calculate the probability that the ensemble is correct using the binomial distribution.
from scipy.stats import binom
# The ensemble is correct if at least 3 out of 5 models are correct.
k = 3  # number of correct models needed
n = number_of_models  # total number of models
p = accuracy_of_each_model  # probability of each model being correct
prob_ensemble_correct = binom.pmf(k, n, p)  + binom.pmf(k + 1, n, p) + binom.pmf(k + 2, n, p)  # sum of probabilities for 3, 4, and 5 correct models    
print(f"Probability that the ensemble is correct: {prob_ensemble_correct:.4f}")

```

```python
Model_1_accuracy = 0.70
Model_2_accuracy = 0.75
Model_3_accuracy = 0.80

from itertools import combinations
from math import prod

def ensemble_accuracy_different(ps):
    n = len(ps)
    majority = n // 2 + 1
    total_prob = 0

    model_indices = range(n)

    for k in range(majority, n + 1):
        for correct_models in combinations(model_indices, k):
            prob = 1

            for i in model_indices:
                if i in correct_models:
                    prob *= ps[i]
                else:
                    prob *= 1 - ps[i]

            total_prob += prob

    return total_prob

accuracies = [0.70, 0.75, 0.80]

print(ensemble_accuracy_different(accuracies))

```

Independent Training Sets

Suppose we collect multiple independent training sets T r1, ..., T rK and use each of these
training sets to train a different instance of the same classifier obtaining K classification
functions f1(x), ..., fK(x).

Classifiers trained in this way are guaranteed to make independent errors on test data.

Bootstrap aggregation or Bagging is an approximation to the previous method that takes
a single training set T r and randomly sub-samples from it K times (with replacement) to
form K training sets T r1, ..., T rK.

Each of these training sets is used to train a different instance of the same classifier
obtaining K classification functions f1(x), ..., fK(x).

The errors won’t by totally independent because the data sets aren’t independent,
but the random re-sampling usually introduces enough diversity to decrease the variance
and give improved performance.
